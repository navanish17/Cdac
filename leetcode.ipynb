```python
#check palindrome without reverse

n = 121

num = n
result = 0
while num > 0:
    id = num%10
    result = result*10 + id
    num = num//10
n == result
```

```python
n = [1,7,3,6,5,6]

for i in range(len(n)):
    ls = sum(n[0:i])
    rs = sum(n[i+1:])
    if ls == rs:
        print(i)
        break
else:
    print("-1")
```

```python
#optimal solution
from math import sqrt
n = 16
s = []
for i in range(1,int(sqrt(n))+1):
    if n%i==0:
        s.append(i)
        if i != (n//i):
            s.append(n//i)
print(s)
```

```python
#frequency mapping
n = [1,7,3,6,5,6,2,1,1,6,7]

f_map = {}

for i in range(len(n)):
    f_map[n[i]] = f_map.get(n[i],0)+1
print(f_map)



```
