```python
"""1. Calculate factorial"""

def factorial(n):
    if n == 1:
        return 1
    return n*factorial(n-1)

factorial(8)
```

```python
"""2. Sum of Digits
Use recursion to find the sum of digits of a number.
Example: 1234 → 1 + 2 + 3 + 4 = 10"""

def sum1(n):
    s = list(map(int, str(n)))
    s = sum(s)
    return s
```
